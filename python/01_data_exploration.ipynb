```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("../data/train.csv")

# Basic information
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

# First 5 rows
display(df.head())

# Data types and missing values
print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

# Duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# Basic statistics
display(df.describe().T)
```

```python
# Data Quality Analysis

# 1. Missing values
missing = df.isnull().sum()
print("Columns with missing values:")
print(missing[missing > 0])

# 2. Duplicate Unique IDs
print("\nDuplicate UniqueIDs:", df["UniqueID"].duplicated().sum())

# 3. Target distribution
print("\nLoan Default Distribution:")
print(df["loan_default"].value_counts())

print("\nLoan Default Percentage:")
print((df["loan_default"].value_counts(normalize=True) * 100).round(2))

# 4. Employment type
print("\nEmployment Type:")
print(df["Employment.Type"].value_counts(dropna=False))

# 5. Credit score description
print("\nCredit Score Description:")
print(df["PERFORM_CNS.SCORE.DESCRIPTION"].value_counts())

# 6. Date range
print("\nDisbursal Date Range:")
print(df["DisbursalDate"].min(), "to", df["DisbursalDate"].max())
```

```python
# ============================================================
# Data Cleaning & Feature Preparation
# ============================================================

# Create a working copy
df_clean = df.copy()

# 1. Handle missing employment type
df_clean["Employment.Type"] = df_clean["Employment.Type"].fillna("Unknown")

# 2. Convert dates
df_clean["Date.of.Birth"] = pd.to_datetime(
    df_clean["Date.of.Birth"],
    format="%d-%m-%y",
    errors="coerce"
)

df_clean["DisbursalDate"] = pd.to_datetime(
    df_clean["DisbursalDate"],
    format="%d-%m-%y",
    errors="coerce"
)

# 3. Create useful loan/customer features
df_clean["Loan_to_Asset_Ratio"] = (
    df_clean["disbursed_amount"] / df_clean["asset_cost"]
).round(4)

df_clean["Loan_Amount_Band"] = pd.cut(
    df_clean["disbursed_amount"],
    bins=[0, 30000, 50000, 75000, 100000, float("inf")],
    labels=[
        "Below 30K",
        "30K-50K",
        "50K-75K",
        "75K-100K",
        "Above 100K"
    ]
)

df_clean["LTV_Band"] = pd.cut(
    df_clean["ltv"],
    bins=[0, 60, 70, 80, 90, 100],
    labels=[
        "Below 60%",
        "60%-70%",
        "70%-80%",
        "80%-90%",
        "90%+"
    ]
)

# 4. Credit bureau availability flag
df_clean["Has_Credit_History"] = np.where(
    df_clean["PERFORM_CNS.SCORE"] > 0,
    "Yes",
    "No"
)

# 5. Previous overdue account flag
df_clean["Has_Previous_Overdue"] = np.where(
    df_clean["PRI.OVERDUE.ACCTS"] > 0,
    "Yes",
    "No"
)

# 6. Recent delinquency flag
df_clean["Recent_Delinquency"] = np.where(
    df_clean["DELINQUENT.ACCTS.IN.LAST.SIX.MONTHS"] > 0,
    "Yes",
    "No"
)

# 7. Verify cleaning
print("Original rows :", len(df))
print("Cleaned rows  :", len(df_clean))

print("\nMissing Employment Type:")
print(df_clean["Employment.Type"].value_counts())

print("\nNew Features:")
print([
    "Loan_to_Asset_Ratio",
    "Loan_Amount_Band",
    "LTV_Band",
    "Has_Credit_History",
    "Has_Previous_Overdue",
    "Recent_Delinquency"
])

print("\nRemaining Missing Values:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python
# ============================================================
# Default Risk Analysis
# ============================================================

# 1. Default rate by Employment Type
employment_risk = (
    df_clean.groupby("Employment.Type")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

employment_risk["default_rate"] = (
    employment_risk["mean"] * 100
).round(2)

print("Default Rate by Employment Type")
display(employment_risk[
    ["Employment.Type", "count", "default_rate"]
])


# 2. Default rate by LTV Band
ltv_risk = (
    df_clean.groupby("LTV_Band", observed=True)["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

ltv_risk["default_rate"] = (
    ltv_risk["mean"] * 100
).round(2)

print("\nDefault Rate by LTV Band")
display(ltv_risk[
    ["LTV_Band", "count", "default_rate"]
])


# 3. Default rate by Credit History
credit_risk = (
    df_clean.groupby("Has_Credit_History")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

credit_risk["default_rate"] = (
    credit_risk["mean"] * 100
).round(2)

print("\nDefault Rate by Credit History")
display(credit_risk[
    ["Has_Credit_History", "count", "default_rate"]
])


# 4. Default rate by Previous Overdue
overdue_risk = (
    df_clean.groupby("Has_Previous_Overdue")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

overdue_risk["default_rate"] = (
    overdue_risk["mean"] * 100
).round(2)

print("\nDefault Rate by Previous Overdue")
display(overdue_risk[
    ["Has_Previous_Overdue", "count", "default_rate"]
])


# 5. Default rate by Recent Delinquency
delinquency_risk = (
    df_clean.groupby("Recent_Delinquency")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

delinquency_risk["default_rate"] = (
    delinquency_risk["mean"] * 100
).round(2)

print("\nDefault Rate by Recent Delinquency")
display(delinquency_risk[
    ["Recent_Delinquency", "count", "default_rate"]
])
```

```python
import matplotlib.pyplot as plt

# 1. Default Rate by Employment Type
plt.figure(figsize=(8, 5))
plt.bar(
    employment_risk["Employment.Type"],
    employment_risk["default_rate"]
)
plt.title("Default Rate by Employment Type")
plt.ylabel("Default Rate (%)")
plt.xlabel("Employment Type")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()


# 2. Default Rate by LTV Band
plt.figure(figsize=(8, 5))
plt.bar(
    ltv_risk["LTV_Band"].astype(str),
    ltv_risk["default_rate"]
)
plt.title("Default Rate by LTV Band")
plt.ylabel("Default Rate (%)")
plt.xlabel("LTV Band")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()


# 3. Default Rate by Credit History
plt.figure(figsize=(7, 5))
plt.bar(
    credit_risk["Has_Credit_History"],
    credit_risk["default_rate"]
)
plt.title("Default Rate by Credit History")
plt.ylabel("Default Rate (%)")
plt.xlabel("Credit History")
plt.tight_layout()
plt.show()


# 4. Default Rate by Previous Overdue
plt.figure(figsize=(7, 5))
plt.bar(
    overdue_risk["Has_Previous_Overdue"],
    overdue_risk["default_rate"]
)
plt.title("Default Rate by Previous Overdue")
plt.ylabel("Default Rate (%)")
plt.xlabel("Previous Overdue")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Loan & Credit Risk Analysis
# ============================================================

# 1. Default rate by Loan Amount Band
loan_risk = (
    df_clean.groupby("Loan_Amount_Band", observed=True)["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

loan_risk["default_rate"] = (loan_risk["mean"] * 100).round(2)

print("Default Rate by Loan Amount")
display(loan_risk[["Loan_Amount_Band", "count", "default_rate"]])


# 2. Default rate by Credit Score
credit_score_risk = (
    df_clean.groupby("PERFORM_CNS.SCORE")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

credit_score_risk["default_rate"] = (
    credit_score_risk["mean"] * 100
).round(2)

print("\nCredit Score Summary")
display(
    credit_score_risk.sort_values(
        "count", ascending=False
    ).head(15)
)


# 3. Default rate by Previous Account History
account_risk = (
    df_clean.groupby("PRI.NO.OF.ACCTS")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

account_risk["default_rate"] = (
    account_risk["mean"] * 100
).round(2)

print("\nPrevious Account Summary")
display(
    account_risk[account_risk["count"] >= 100]
    .sort_values("count", ascending=False)
    .head(15)
)


# 4. Default rate by Number of Credit Inquiries
inquiry_risk = (
    df_clean.groupby("NO.OF_INQUIRIES")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

inquiry_risk["default_rate"] = (
    inquiry_risk["mean"] * 100
).round(2)

print("\nDefault Rate by Credit Inquiries")
display(inquiry_risk)
```

```python
# ============================================================
# Customer Risk Profile
# ============================================================

df_clean["Risk_Score"] = (
    (df_clean["ltv"] >= 80).astype(int)
    + (df_clean["PRI.OVERDUE.ACCTS"] > 0).astype(int)
    + (df_clean["DELINQUENT.ACCTS.IN.LAST.SIX.MONTHS"] > 0).astype(int)
    + (df_clean["NO.OF_INQUIRIES"] >= 2).astype(int)
)

# Create risk categories
df_clean["Risk_Category"] = pd.cut(
    df_clean["Risk_Score"],
    bins=[-1, 0, 1, 2, 4],
    labels=["Low", "Moderate", "High", "Very High"]
)

# Analyze default rate by risk category
risk_analysis = (
    df_clean.groupby("Risk_Category", observed=True)["loan_default"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

risk_analysis["default_rate"] = (
    risk_analysis["mean"] * 100
).round(2)

print("Customer Risk Profile")
display(
    risk_analysis[
        ["Risk_Category", "count", "sum", "default_rate"]
    ]
)

# Distribution of customers across risk categories
print("\nCustomer Distribution:")
display(
    df_clean["Risk_Category"]
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# Geographic & Operational Risk Analysis
# ============================================================

# 1. Default rate by State
state_risk = (
    df_clean.groupby("State_ID")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

state_risk["default_rate"] = (
    state_risk["mean"] * 100
).round(2)

print("Default Rate by State")
display(
    state_risk
    .sort_values("default_rate", ascending=False)
)


# 2. Branch-level risk
branch_risk = (
    df_clean.groupby("branch_id")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

branch_risk["default_rate"] = (
    branch_risk["mean"] * 100
).round(2)

print("\nHighest-Risk Branches")
display(
    branch_risk[branch_risk["count"] >= 500]
    .sort_values("default_rate", ascending=False)
    .head(10)
)


# 3. Manufacturer-level risk
manufacturer_risk = (
    df_clean.groupby("manufacturer_id")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

manufacturer_risk["default_rate"] = (
    manufacturer_risk["mean"] * 100
).round(2)

print("\nManufacturer Risk")
display(
    manufacturer_risk[manufacturer_risk["count"] >= 500]
    .sort_values("default_rate", ascending=False)
    .head(10)
)


# 4. Supplier-level risk
supplier_risk = (
    df_clean.groupby("supplier_id")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

supplier_risk["default_rate"] = (
    supplier_risk["mean"] * 100
).round(2)

print("\nHighest-Risk Suppliers")
display(
    supplier_risk[supplier_risk["count"] >= 500]
    .sort_values("default_rate", ascending=False)
    .head(10)
)
```

```python
# ============================================================
# Geographic & Operational Risk Analysis
# ============================================================

# 1. Default rate by State
state_risk = (
    df_clean.groupby("State_ID")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

state_risk["default_rate"] = (
    state_risk["mean"] * 100
).round(2)

print("Default Rate by State")
display(
    state_risk
    .sort_values("default_rate", ascending=False)
)


# 2. Branch-level risk
branch_risk = (
    df_clean.groupby("branch_id")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

branch_risk["default_rate"] = (
    branch_risk["mean"] * 100
).round(2)

print("\nHighest-Risk Branches")
display(
    branch_risk[branch_risk["count"] >= 500]
    .sort_values("default_rate", ascending=False)
    .head(10)
)


# 3. Manufacturer-level risk
manufacturer_risk = (
    df_clean.groupby("manufacturer_id")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

manufacturer_risk["default_rate"] = (
    manufacturer_risk["mean"] * 100
).round(2)

print("\nManufacturer Risk")
display(
    manufacturer_risk[manufacturer_risk["count"] >= 500]
    .sort_values("default_rate", ascending=False)
    .head(10)
)


# 4. Supplier-level risk
supplier_risk = (
    df_clean.groupby("supplier_id")["loan_default"]
    .agg(["count", "mean"])
    .reset_index()
)

supplier_risk["default_rate"] = (
    supplier_risk["mean"] * 100
).round(2)

print("\nHighest-Risk Suppliers")
display(
    supplier_risk[supplier_risk["count"] >= 500]
    .sort_values("default_rate", ascending=False)
    .head(10)
)
```

```python
# ============================================================
# High-Risk Supplier & Branch Deep Dive
# ============================================================

# High-risk suppliers
high_risk_suppliers = supplier_risk[
    supplier_risk["count"] >= 500
].sort_values(
    "default_rate", ascending=False
).head(10)

print("High-Risk Suppliers")
display(high_risk_suppliers)


# Compare high-risk suppliers with overall portfolio
overall_default = df_clean["loan_default"].mean() * 100

high_risk_suppliers["vs_portfolio"] = (
    high_risk_suppliers["default_rate"] - overall_default
).round(2)

print("\nSupplier Default Rate vs Overall Portfolio")
display(
    high_risk_suppliers[
        ["supplier_id", "count", "default_rate", "vs_portfolio"]
    ]
)


# High-risk branches
high_risk_branches = branch_risk[
    branch_risk["count"] >= 500
].sort_values(
    "default_rate", ascending=False
).head(10)

print("\nHigh-Risk Branches")
display(
    high_risk_branches[
        ["branch_id", "count", "default_rate"]
    ]
)


# Overall portfolio benchmark
print(
    f"\nOverall Portfolio Default Rate: "
    f"{overall_default:.2f}%"
)
```

```python
# ============================================================
# Monthly Loan & Default Trend
# ============================================================

# Create month from disbursal date
df_clean["Disbursal_Month"] = df_clean["DisbursalDate"].dt.to_period("M")

monthly_risk = (
    df_clean.groupby("Disbursal_Month")["loan_default"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

monthly_risk["default_rate"] = (
    monthly_risk["mean"] * 100
).round(2)

monthly_risk["default_count"] = monthly_risk["sum"]

print("Monthly Portfolio Performance")
display(
    monthly_risk[
        ["Disbursal_Month", "count", "default_count", "default_rate"]
    ]
)

# Monthly loan amount
monthly_amount = (
    df_clean.groupby("Disbursal_Month")["disbursed_amount"]
    .agg(["sum", "mean"])
    .reset_index()
)

monthly_amount.columns = [
    "Disbursal_Month",
    "Total_Disbursed",
    "Average_Loan"
]

print("\nMonthly Loan Amount")
display(monthly_amount)
```

```python
# ============================================================
# Create Final Analytical Dataset
# ============================================================

analysis_df = df_clean[
    [
        "UniqueID",
        "disbursed_amount",
        "asset_cost",
        "ltv",
        "Employment.Type",
        "DisbursalDate",
        "State_ID",
        "branch_id",
        "manufacturer_id",
        "supplier_id",
        "PERFORM_CNS.SCORE",
        "Has_Credit_History",
        "PRI.NO.OF.ACCTS",
        "PRI.OVERDUE.ACCTS",
        "DELINQUENT.ACCTS.IN.LAST.SIX.MONTHS",
        "NO.OF_INQUIRIES",
        "Loan_Amount_Band",
        "LTV_Band",
        "Has_Previous_Overdue",
        "Recent_Delinquency",
        "Risk_Score",
        "Risk_Category",
        "loan_default"
    ]
].copy()

# Save analytical dataset
analysis_df.to_csv(
    "../data/vehicle_loan_analysis.csv",
    index=False
)

print("Final analytical dataset created.")
print("Rows:", len(analysis_df))
print("Columns:", len(analysis_df.columns))

display(analysis_df.head())
```
